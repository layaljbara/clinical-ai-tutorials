## 1. The Problem: Why Generators Exist

### The Memory Wall

Consider this common scenario:

```python
# Processing 1 million records
def get_all_user_ids(database):
    user_ids = []
    for record in database.fetch_all():  # Loads ALL records into memory
        user_ids.append(record['id'])
    return user_ids

# What if you only need the first 10?
first_ten = get_all_user_ids(database)[:10]  # Wasteful!
