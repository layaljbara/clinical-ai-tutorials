## Introduction to OOP

**Object-Oriented Programming (OOP)** is a programming paradigm that organizes code around **objects** rather than functions and logic. An object is a bundle of data (attributes) and behaviors (methods).

### Four Pillars of OOP:
1. **Encapsulation**: Bundling data and methods that operate on that data within a single unit (class)
2. **Abstraction**: Hiding complex implementation details and showing only necessary features
3. **Inheritance**: Creating new classes based on existing classes
4. **Polymorphism**: Using a single interface to represent different types

## Classes and Objects

### What is a Class?
A **class** is a blueprint or template for creating objects. It defines the structure and behavior that objects of that type will have.

### What is an Object?
An **object** is an instance of a class. It's a concrete realization of the class blueprint.

### Basic Class Definition

```python
# Define a simple class
class Dog:
    pass  # Empty class for now

# Create an object (instance) of the Dog class
my_dog = Dog()
print(type(my_dog))  # Output: <class '__main__.Dog'>
```

## The `__init__` Method (Constructor)

The `__init__` method is a special method called a **constructor**. It's automatically called when you create a new instance of a class.

```python
class Dog:
    def __init__(self, name, age):
        """Constructor method - initializes the object"""
        self.name = name  # Instance variable
        self.age = age    # Instance variable

# Create instances
dog1 = Dog("Buddy", 3)
dog2 = Dog("Max", 5)

print(dog1.name)  # Output: Buddy
print(dog2.age)   # Output: 5
```

### Key Points:
- `__init__` is called automatically when creating an object
- The first parameter is always `self` (refers to the instance being created)
- You can have additional parameters to initialize the object's state

## Instance Variables and Methods

### Instance Variables
Instance variables are attributes that belong to a specific instance of a class. Each object has its own copy.

```python
class Person:
    def __init__(self, name, age):
        self.name = name      # Instance variable
        self.age = age        # Instance variable
        self.hobbies = []     # Instance variable

person1 = Person("Alice", 30)
person2 = Person("Bob", 25)

person1.hobbies.append("Reading")
person2.hobbies.append("Gaming")

print(person1.hobbies)  # Output: ['Reading']
print(person2.hobbies)  # Output: ['Gaming']
```

### Instance Methods
Instance methods are functions defined inside a class that operate on instance data.

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def bark(self):
        """Instance method"""
        return f"{self.name} says Woof!"
    
    def get_age_in_dog_years(self):
        """Instance method that performs a calculation"""
        return self.age * 7
    
    def celebrate_birthday(self):
        """Instance method that modifies instance data"""
        self.age += 1
        return f"Happy Birthday {self.name}! Now {self.age} years old."

# Using instance methods
my_dog = Dog("Buddy", 3)
print(my_dog.bark())                    # Output: Buddy says Woof!
print(my_dog.get_age_in_dog_years())    # Output: 21
print(my_dog.celebrate_birthday())      # Output: Happy Birthday Buddy! Now 4 years old.
```

## The `self` Parameter

`self` represents the instance of the class. It's used to access variables and methods associated with the current object.

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius  # self.radius is an instance variable
    
    def get_area(self):
        # self is used to access the instance variable
        return 3.14159 * self.radius ** 2
    
    def get_circumference(self):
        # self is used to access the instance variable
        return 2 * 3.14159 * self.radius
    
    def scale(self, factor):
        # self is used to modify the instance variable
        self.radius *= factor

circle = Circle(5)
print(circle.get_area())         # Output: 78.53975
circle.scale(2)                  # Doubles the radius
print(circle.get_area())         # Output: 314.159
```

### Important Notes:
- `self` is NOT a keyword (you could use another name, but `self` is the convention)
- `self` must be the first parameter of instance methods
- Python automatically passes the instance as the first argument

## Class Variables vs Instance Variables

### Class Variables
Class variables are shared among all instances of a class.

```python
class Employee:
    # Class variable - shared by all instances
    company_name = "TechCorp"
    employee_count = 0
    
    def __init__(self, name, salary):
        # Instance variables - unique to each instance
        self.name = name
        self.salary = salary
        Employee.employee_count += 1  # Modify class variable

# All instances share the same class variable
emp1 = Employee("Alice", 50000)
emp2 = Employee("Bob", 60000)

print(emp1.company_name)        # Output: TechCorp
print(emp2.company_name)        # Output: TechCorp
print(Employee.employee_count)  # Output: 2

# Changing class variable affects all instances
Employee.company_name = "NewTech"
print(emp1.company_name)        # Output: NewTech
print(emp2.company_name)        # Output: NewTech
```

### Key Differences:

| Feature | Instance Variable | Class Variable |
|---------|------------------|----------------|
| Definition | Inside `__init__` with `self` | Inside class, outside methods |
| Access | `self.variable_name` | `ClassName.variable_name` |
| Scope | Unique to each instance | Shared across all instances |
| Memory | Separate copy per instance | Single copy for all instances |

## Instance Methods, Class Methods, and Static Methods

### Instance Methods
Regular methods that operate on instance data (require `self`).

```python
class Calculator:
    def __init__(self, value):
        self.value = value
    
    def add(self, x):
        """Instance method - uses self"""
        self.value += x
        return self.value
```

### Class Methods
Methods that operate on class-level data (use `@classmethod` decorator and `cls` parameter).

```python
class Person:
    population = 0  # Class variable
    
    def __init__(self, name):
        self.name = name
        Person.population += 1
    
    @classmethod
    def get_population(cls):
        """Class method - uses cls instead of self"""
        return f"Total population: {cls.population}"
    
    @classmethod
    def create_anonymous(cls):
        """Class method can be used as alternative constructor"""
        return cls("Anonymous")

# Using class methods
person1 = Person("Alice")
person2 = Person("Bob")
print(Person.get_population())      # Output: Total population: 2

# Alternative constructor
anon = Person.create_anonymous()
print(anon.name)                    # Output: Anonymous
```

### Static Methods
Methods that don't operate on instance or class data (use `@staticmethod` decorator).

```python
class MathUtils:
    @staticmethod
    def add(x, y):
        """Static method - no self or cls parameter"""
        return x + y
    
    @staticmethod
    def is_even(number):
        """Static method - utility function"""
        return number % 2 == 0

# Using static methods (no instance needed)
print(MathUtils.add(5, 3))          # Output: 8
print(MathUtils.is_even(4))         # Output: True
```

### Comparison Table:

| Method Type | Decorator | First Parameter | Access to | Use Case |
|-------------|-----------|-----------------|-----------|----------|
| Instance | None | `self` | Instance & class data | Operations on instance data |
| Class | `@classmethod` | `cls` | Class data only | Factory methods, class-level operations |
| Static | `@staticmethod` | None | Neither | Utility functions related to class |

## Encapsulation and Access Modifiers

Encapsulation is the practice of hiding internal details and providing a public interface.

### Python's Access Levels:

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner              # Public attribute
        self._account_number = "12345"  # Protected attribute (convention)
        self.__balance = balance        # Private attribute (name mangling)
    
    # Public method
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            return True
        return False
    
    # Public method
    def get_balance(self):
        """Getter method for private attribute"""
        return self.__balance
    
    # Protected method (convention)
    def _generate_statement(self):
        return f"Balance: ${self.__balance}"
    
    # Private method
    def __validate_transaction(self, amount):
        return amount > 0 and amount <= self.__balance

account = BankAccount("Alice", 1000)

# Public access
print(account.owner)            # Output: Alice

# Protected access (still accessible, but convention says don't)
print(account._account_number)  # Output: 12345

# Private access (name mangling makes it harder to access)
# print(account.__balance)      # AttributeError
print(account.get_balance())    # Output: 1000 (proper way)

# Accessing private attribute (not recommended)
print(account._BankAccount__balance)  # Output: 1000 (name mangling)
```

### Naming Conventions:

| Convention | Meaning | Example |
|------------|---------|---------|
| `name` | Public | `self.name` |
| `_name` | Protected (internal use) | `self._internal_id` |
| `__name` | Private (name mangling) | `self.__secret` |
| `__name__` | Special/magic methods | `__init__`, `__str__` |


| Method type     | Who calls it       | Who overrides it | Why                  |
| --------------- | ------------------ | ---------------- | -------------------- |
| Public          | Users              | Subclasses       | API                  |
| Protected (`_`) | Class & subclasses | Subclasses       | Extensible internals |
| Private (`__`)  | Class only         | Nobody           | Safety & invariants  |

## Inheritance

Inheritance allows a class to inherit attributes and methods from another class.

### Basic Inheritance

```python
# Parent class (Base class, Superclass)
class Animal:
    def __init__(self, name, species):
        self.name = name
        self.species = species
    
    def make_sound(self):
        return "Some generic sound"
    
    def info(self):
        return f"{self.name} is a {self.species}"

# Child class (Derived class, Subclass)
class Dog(Animal):
    def __init__(self, name, breed):
        # Call parent constructor
        super().__init__(name, "Dog")
        self.breed = breed
    
    # Override parent method
    def make_sound(self):
        return "Woof!"
    
    # Add new method specific to Dog
    def fetch(self):
        return f"{self.name} is fetching the ball!"

# Using inheritance
dog = Dog("Buddy", "Golden Retriever")
print(dog.info())           # Output: Buddy is a Dog (inherited method)
print(dog.make_sound())     # Output: Woof! (overridden method)
print(dog.fetch())          # Output: Buddy is fetching the ball! (new method)
```

### Multiple Inheritance

```python
class Flyer:
    def fly(self):
        return "Flying high!"

class Swimmer:
    def swim(self):
        return "Swimming fast!"

class Duck(Animal, Flyer, Swimmer):
    def __init__(self, name):
        super().__init__(name, "Duck")
    
    def make_sound(self):
        return "Quack!"

duck = Duck("Donald")
print(duck.make_sound())    # Output: Quack!
print(duck.fly())           # Output: Flying high!
print(duck.swim())          # Output: Swimming fast!
```

### Method Resolution Order (MRO)

```python
class A:
    def method(self):
        return "A"

class B(A):
    def method(self):
        return "B"

class C(A):
    def method(self):
        return "C"

class D(B, C):
    pass

d = D()
print(d.method())           # Output: B
print(D.mro())              # Shows method resolution order
# Output: [<class '__main__.D'>, <class '__main__.B'>, <class '__main__.C'>, <class '__main__.A'>, <class 'object'>]
```

## Polymorphism

Polymorphism allows objects of different classes to be treated as objects of a common base class.

### Method Overriding

```python
class Shape:
    def area(self):
        pass
    
    def perimeter(self):
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height
    
    def perimeter(self):
        return 2 * (self.width + self.height)

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14159 * self.radius ** 2
    
    def perimeter(self):
        return 2 * 3.14159 * self.radius

# Polymorphism in action
shapes = [Rectangle(5, 10), Circle(7), Rectangle(3, 4)]

for shape in shapes:
    # Same method call, different behavior based on object type
    print(f"Area: {shape.area():.2f}, Perimeter: {shape.perimeter():.2f}")
```

### Duck Typing

Python uses "duck typing" - if it walks like a duck and quacks like a duck, it's a duck.

```python
class Dog:
    def speak(self):
        return "Woof!"

class Cat:
    def speak(self):
        return "Meow!"

class Robot:
    def speak(self):
        return "Beep boop!"

def make_it_speak(thing):
    """Works with any object that has a speak() method"""
    return thing.speak()

# All different types, but all have speak() method
print(make_it_speak(Dog()))     # Output: Woof!
print(make_it_speak(Cat()))     # Output: Meow!
print(make_it_speak(Robot()))   # Output: Beep boop!
```

## Special Methods (Magic Methods)

Special methods (also called dunder methods) allow you to define how objects behave with built-in operations.

### Common Special Methods

```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
    
    def __str__(self):
        """String representation for users (print, str())"""
        return f"'{self.title}' by {self.author}"
    
    def __repr__(self):
        """String representation for developers (repr())"""
        return f"Book('{self.title}', '{self.author}', {self.pages})"
    
    def __len__(self):
        """Define behavior for len()"""
        return self.pages
    
    def __eq__(self, other):
        """Define behavior for == operator"""
        if isinstance(other, Book):
            return self.title == other.title and self.author == other.author
        return False
    
    def __lt__(self, other):
        """Define behavior for < operator"""
        return self.pages < other.pages
    
    def __add__(self, other):
        """Define behavior for + operator"""
        if isinstance(other, Book):
            return self.pages + other.pages
        return NotImplemented

# Using special methods
book1 = Book("1984", "George Orwell", 328)
book2 = Book("Animal Farm", "George Orwell", 112)

print(book1)                    # Output: '1984' by George Orwell (__str__)
print(repr(book1))              # Output: Book('1984', 'George Orwell', 328) (__repr__)
print(len(book1))               # Output: 328 (__len__)
print(book1 == book2)           # Output: False (__eq__)
print(book2 < book1)            # Output: True (__lt__)
print(book1 + book2)            # Output: 440 (__add__)
```

### More Special Methods

```python
class Counter:
    def __init__(self, start=0):
        self.value = start
    
    def __call__(self):
        """Makes the object callable like a function"""
        self.value += 1
        return self.value
    
    def __getitem__(self, key):
        """Define behavior for indexing: obj[key]"""
        return self.value * key
    
    def __contains__(self, item):
        """Define behavior for 'in' operator"""
        return item == self.value

counter = Counter(5)
print(counter())                # Output: 6 (callable)
print(counter[3])               # Output: 18 (6 * 3)
print(6 in counter)             # Output: True
```

### Useful Special Methods Reference:

| Method | Operator/Function | Description |
|--------|------------------|-------------|
| `__init__` | Constructor | Initialize object |
| `__str__` | `str()`, `print()` | User-friendly string |
| `__repr__` | `repr()` | Developer-friendly string |
| `__len__` | `len()` | Length of object |
| `__getitem__` | `obj[key]` | Indexing |
| `__setitem__` | `obj[key] = value` | Assignment |
| `__eq__` | `==` | Equality |
| `__lt__` | `<` | Less than |
| `__le__` | `<=` | Less than or equal |
| `__gt__` | `>` | Greater than |
| `__add__` | `+` | Addition |
| `__sub__` | `-` | Subtraction |
| `__mul__` | `*` | Multiplication |
| `__call__` | `obj()` | Make object callable |
| `__contains__` | `in` | Membership test |

## Property Decorators

Properties allow you to use methods as if they were attributes, providing controlled access.

### Using `@property`

```python
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius
    
    @property
    def celsius(self):
        """Getter for celsius"""
        return self._celsius
    
    @celsius.setter
    def celsius(self, value):
        """Setter for celsius with validation"""
        if value < -273.15:
            raise ValueError("Temperature below absolute zero is not possible")
        self._celsius = value
    
    @property
    def fahrenheit(self):
        """Computed property"""
        return self._celsius * 9/5 + 32
    
    @fahrenheit.setter
    def fahrenheit(self, value):
        """Set celsius from fahrenheit"""
        self.celsius = (value - 32) * 5/9

# Using properties
temp = Temperature(25)
print(temp.celsius)         # Output: 25 (calls getter)
print(temp.fahrenheit)      # Output: 77.0 (computed)

temp.celsius = 30           # Calls setter
print(temp.celsius)         # Output: 30

temp.fahrenheit = 86        # Sets celsius via fahrenheit
print(temp.celsius)         # Output: 30.0

# temp.celsius = -300       # Raises ValueError
```

### Read-Only Properties

```python
class Circle:
    def __init__(self, radius):
        self._radius = radius
    
    @property
    def radius(self):
        return self._radius
    
    @property
    def area(self):
        """Read-only computed property"""
        return 3.14159 * self._radius ** 2
    
    @property
    def diameter(self):
        """Read-only computed property"""
        return self._radius * 2

circle = Circle(5)
print(circle.area)          # Output: 78.53975
print(circle.diameter)      # Output: 10
# circle.area = 100         # AttributeError (no setter defined)
```

## Best Practices

### 1. Use Clear and Descriptive Names

```python
# Good
class CustomerAccount:
    def calculate_total_balance(self):
        pass

# Avoid
class CA:
    def calc(self):
        pass
```

### 2. Follow the Single Responsibility Principle

```python
# Good - Each class has one responsibility
class User:
    def __init__(self, username, email):
        self.username = username
        self.email = email

class UserValidator:
    @staticmethod
    def validate_email(email):
        return "@" in email

class UserRepository:
    def save(self, user):
        # Save to database
        pass

# Avoid - Too many responsibilities
class User:
    def __init__(self, username, email):
        self.username = username
        self.email = email
    
    def validate_email(self):
        pass
    
    def save_to_database(self):
        pass
    
    def send_welcome_email(self):
        pass
```

### 3. Use Composition Over Inheritance When Appropriate

```python
# Composition - "has-a" relationship
class Engine:
    def start(self):
        return "Engine started"

class Car:
    def __init__(self):
        self.engine = Engine()  # Car HAS an Engine
    
    def start(self):
        return self.engine.start()

# Inheritance - "is-a" relationship
class Vehicle:
    pass

class Car(Vehicle):  # Car IS a Vehicle
    pass
```

### 5. Use Docstrings

```python
class BankAccount:
    """
    A class representing a bank account.
    
    Attributes:
        owner (str): The name of the account owner
        balance (float): The current account balance
    """
    
    def __init__(self, owner: str, initial_balance: float = 0):
        """
        Initialize a new bank account.
        
        Args:
            owner: The name of the account owner
            initial_balance: Starting balance (default: 0)
        """
        self.owner = owner
        self.balance = initial_balance
    
    def deposit(self, amount: float) -> bool:
        """
        Deposit money into the account.
        
        Args:
            amount: The amount to deposit
        
        Returns:
            True if successful, False otherwise
        """
        if amount > 0:
            self.balance += amount
            return True
        return False
```

### 6. Avoid Mutable Default Arguments

```python
# Bad - Mutable default argument
class Team:
    def __init__(self, members=[]):  # Don't do this!
        self.members = members

# Good - Use None and create new list
class Team:
    def __init__(self, members=None):
        self.members = members if members is not None else []
```

### 7. Use `__repr__` for Debugging

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __repr__(self):
        return f"Point({self.x}, {self.y})"
    
    def __str__(self):
        return f"({self.x}, {self.y})"

p = Point(3, 4)
print(p)            # Output: (3, 4) - uses __str__
print(repr(p))      # Output: Point(3, 4) - uses __repr__
print([p])          # Output: [Point(3, 4)] - uses __repr__ in containers
```

## Complete Example: Putting It All Together

```python
from typing import List, Optional
from datetime import datetime

class Book:
    """Represents a book in a library."""
    
    # Class variable
    total_books = 0
    
    def __init__(self, title: str, author: str, isbn: str, pages: int):
        """Initialize a new book."""
        self.title = title
        self.author = author
        self.isbn = isbn
        self._pages = pages
        self._is_available = True
        Book.total_books += 1
    
    @property
    def pages(self) -> int:
        """Get the number of pages."""
        return self._pages
    
    @property
    def is_available(self) -> bool:
        """Check if book is available."""
        return self._is_available
    
    def checkout(self) -> bool:
        """Checkout the book."""
        if self._is_available:
            self._is_available = False
            return True
        return False
    
    def return_book(self) -> None:
        """Return the book."""
        self._is_available = True
    
    def __str__(self) -> str:
        return f"'{self.title}' by {self.author}"
    
    def __repr__(self) -> str:
        return f"Book('{self.title}', '{self.author}', '{self.isbn}', {self._pages})"
    
    def __eq__(self, other) -> bool:
        if isinstance(other, Book):
            return self.isbn == other.isbn
        return False
    
    @classmethod
    def get_total_books(cls) -> int:
        """Get total number of books created."""
        return cls.total_books


class Library:
    """Represents a library containing books."""
    
    def __init__(self, name: str):
        """Initialize a new library."""
        self.name = name
        self._books: List[Book] = []
    
    def add_book(self, book: Book) -> None:
        """Add a book to the library."""
        self._books.append(book)
    
    def find_book(self, title: str) -> Optional[Book]:
        """Find a book by title."""
        for book in self._books:
            if book.title.lower() == title.lower():
                return book
        return None
    
    def get_available_books(self) -> List[Book]:
        """Get all available books."""
        return [book for book in self._books if book.is_available]
    
    def __len__(self) -> int:
        return len(self._books)
    
    def __str__(self) -> str:
        return f"{self.name} Library ({len(self)} books)"


# Using the classes
if __name__ == "__main__":
    # Create library
    library = Library("City")
    
    # Create books
    book1 = Book("1984", "George Orwell", "978-0451524935", 328)
    book2 = Book("To Kill a Mockingbird", "Harper Lee", "978-0061120084", 324)
    book3 = Book("The Great Gatsby", "F. Scott Fitzgerald", "978-0743273565", 180)
    
    # Add books to library
    library.add_book(book1)
    library.add_book(book2)
    library.add_book(book3)
    
    print(library)  # Output: City Library (3 books)
    print(f"Total books created: {Book.get_total_books()}")  # Output: 3
    
    # Checkout a book
    book1.checkout()
    print(f"{book1} available: {book1.is_available}")  # Output: False
    
    # Get available books
    available = library.get_available_books()
    print(f"Available books: {len(available)}")  # Output: 2
    
    # Return book
    book1.return_book()
    print(f"{book1} available: {book1.is_available}")  # Output: True
```
