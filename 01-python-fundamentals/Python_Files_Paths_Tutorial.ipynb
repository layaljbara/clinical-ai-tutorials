# Python Files & Paths Tutorial: `os`, `glob`, and `pathlib`

---

## 1. What is a file path?

A **file path** tells Python where a file or folder lives on your system.

There are two main kinds.

---

### 1.1 Absolute paths

An **absolute path** starts from the root of the system.

Examples:

```python
    /path/to/project/data/file.txt
    /home/user/data/file.txt

```
Absolute paths are unambiguous, but often **not portable** across machines.

---

### 1.2 Relative paths

A **relative path** is relative to where Python is currently running.

Example:

```python
    data/file.txt

```
> Most `FileNotFoundError` bugs come from misunderstanding relative paths.

To check where Python is running from:

```python
    import os
    os.getcwd()
```
---

## 2. The `os` module (classic, low-level)

The `os` module provides **direct access to the operating system**.

Use it when you need:
- system-level control
- environment interaction
- backward compatibility

---

### 2.1 Joining paths safely

Never build paths with string concatenation.

Bad:

```python
    "path/" + "file.txt"

```
Correct:

```python
    import os
    os.path.join("data", "pmc", "file.xml")
```
This works on **Linux, macOS, and Windows**.

---

### 2.2 Checking existence

```python
    os.path.exists("data/file.txt")
    os.path.isfile("data/file.txt")
    os.path.isdir("data")

```
Use these before opening files to avoid crashes.

---

### 2.3 Creating directories

```python
    os.makedirs("results/output", exist_ok=True)
```
- creates parent directories if needed  
- does nothing if the directory already exists  

---

## 3. The `glob` module (pattern-based search)


`glob` lets you say:

> “Give me all files that match this pattern.”

---

### 3.1 Basic usage

```python
    import glob
    files = glob.glob("data/*.xml")
```
This finds **all `.xml` files directly inside `data/`**.

---

### 3.2 Recursive search

    files = glob.glob("data/**/*.xml", recursive=True)

This searches:
- `data/`
- all subfolders
- for `.xml` files

---

### 3.3 What `glob` returns

```python
    list[str]
```
A list of file paths as **strings**.

---

## 4. `pathlib` (modern, recommended)

`pathlib` is the **modern, clean, Pythonic way** to work with files.

> Paths are objects, not strings.

---

### 4.1 Creating and joining paths

```python
    from pathlib import Path
    base = Path("data")
    file_path = base / "pmc" / "article.xml"
```
No `os.path.join`, no string manipulation.

---

### 4.2 Checking paths
```python
    file_path.exists()
    file_path.is_file()
    file_path.is_dir()
```
These replace `os.path.*` checks.

---

### 4.3 Listing files
```python
    list(base.glob("*.xml"))
```
Recursive:
```python
    list(base.rglob("*.xml"))
```
This is the `pathlib` equivalent of `glob.glob`.

---

### 4.4 Reading files
```python
    text = file_path.read_text()
    data = file_path.read_bytes()
```
Cleaner than manual `open()` calls for simple use cases.

---

### 4.5 Creating directories
```python
    Path("results/output").mkdir(parents=True, exist_ok=True)
```
---

## 5. `glob` vs `pathlib.glob`

These two are equivalent:
```python
    glob.glob("data/*.xml")
    list(Path("data").glob("*.xml"))
```
Difference:

- `glob` → quick, string-based
- `pathlib` → clean, object-oriented, recommended

---

## 6. Common real-world pattern
```python
    from pathlib import Path
    data_dir = Path("data/pmc")

    for xml_file in data_dir.rglob("*.xml"):
        print(xml_file)
```
This safely finds **all XML files recursively**.

---

## 7. When to use what

- `pathlib` → modern code, file pipelines
- `glob` → quick pattern matching
- `os` → system-level tasks (env vars, permissions)

Avoid hard-coded absolute paths whenever possible.

---

## 8. Final mental model

- 🟢 `pathlib` = modern, readable paths  
- 🔵 `glob` = pattern-based file discovery  
- 🔴 `os` = low-level system interface
