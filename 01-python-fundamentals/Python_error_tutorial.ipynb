# Python Error Handling Tutorial: `try`, `except`, and `raise`

---

## 1. What is an error (exception)?

An **exception** is Python’s way of saying:

> “I cannot continue running this program normally.”

Example:

```python
int("abc")
```

Python stops and shows:

```
ValueError: invalid literal for int() with base 10: 'abc'
```

Once an exception happens:

* execution stops immediately
* lines below are not run

---

## 2. What does `raise` mean?

### Core rule (very important)

> **`raise` ALWAYS stops execution by triggering an error.**

It does **not** print.
It does **not** continue.
Python only *displays* the error because the program has crashed.

---

### 2.1 `raise ValueError("message")`

```python
raise ValueError("Age must be positive")
```

What happens step by step:

1. Python creates a new `ValueError`
2. The message is attached
3. Program execution stops
4. Python displays the error message

Use this when:

* input values are invalid
* a logical condition is violated
* continuing would produce incorrect results

---

### 2.2 `raise` (by itself)

```python
raise
```

Rules:

* Can only be used **inside an `except` block**
* Re-raises the **same error** that was just caught

Example:

```python
try:
    int("abc")
except ValueError:
    raise
```

Meaning:

* the error is caught
* then immediately thrown again
* program stops

This is useful when you want to log or inspect an error but still crash.

Avoid:
```python
except:
    raise
```

because it catches everything, including system-level exits.


---

### 2.3 `raise NewError(...) from old_error`

```python
try:
    int("abc")
except ValueError as e:
    raise RuntimeError("Parsing failed") from e
```

“At this stage of the program, the relevant fact is not that int() failed —
the relevant fact is that parsing failed, and here is the underlying reason.”

Meaning:

* the original error is preserved
* a higher-level explanation is added. Low-level means “closer to the technical operation.” High-level means “closer to the meaning of the program.”

This is called **exception chaining** and is very useful for debugging.


---

## 3. What does `try / except` mean?

`try / except` allows you to say:

> “Try this code. If it fails, don’t crash immediately — do something else.”

---

### 3.1 Without `try / except` (program crashes)

```python
print("A")
int("abc")
print("B")
```

Output:

```
A
ValueError: invalid literal for int()
```

---

### 3.2 With `try / except` (program continues)

```python
print("A")
try:
    int("abc")
except ValueError:
    print("Conversion failed")
print("B")
```

Output:

```
A
Conversion failed
B
```

---

## 4. `except` catches error TYPES

```python
except ValueError:
```

Means:

> “If a `ValueError` occurs, run this block.”

Valid examples:

```python
except ValueError:
except TypeError:
except Exception:
```

Invalid examples:

```python
except raise      # ❌
except print      # ❌
```

Only **exception types** are allowed after `except`.

---

## 5. Difference between `except ValueError` and `except:`

### 5.1 `except ValueError` (recommended)

```python
try:
    int("abc")
except ValueError:
    print("Bad number")
```

* catches only expected errors
* other bugs crash normally

---

### 5.2 `except Exception` (broad but acceptable)

```python
try:
    risky_code()
except Exception as e:
    print("Something failed:", e)
```

* catches most runtime errors
* does NOT catch system exits or keyboard interrupts

---

### 5.3 `except:` (dangerous)

```python
try:
    risky_code()
except:
    print("Ignored everything")
```

⚠️ Catches **everything**, including:

* programming bugs
* `KeyboardInterrupt`
* `SystemExit`

Usually avoided in good code.

---

## 6. Combining `try / except` with `raise`

### Scenario 1: Catch and STOP

```python
try:
    int("abc")
except ValueError:
    raise
```

* error is caught
* immediately re-raised
* program stops

---

### Scenario 2: Catch and CONTINUE

```python
try:
    int("abc")
except ValueError:
    print("Skipping this")
```

* error is expected
* program continues

---

### Scenario 3: Catch, LOG, CONTINUE (pipelines)

```python
try:
    risky_api_call()
except Exception as e:
    print("Failed:", e)
    continue
```

* failure is acceptable
* move on to next item

---

### Scenario 4: Catch, TRANSLATE, STOP

```python
try:
    parse_xml()
except ET.ParseError as e:
    raise RuntimeError("Invalid XML") from e
```

* low-level error wrapped
* higher-level meaning added
* program stops

---

## 7. Golden rules (memorize these)

* `raise` → STOP the program with an error
* `try / except` → decide what to do when an error happens
* catch **specific errors** whenever possible
* avoid bare `except:` unless absolutely necessary

---

## 8. Final mental model

> 🔴 `raise` = “This must not continue”
> 🔵 `except` = “I know how to handle this error”
