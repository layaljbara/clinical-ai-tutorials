```python
### B) glob finds matching files
import glob, os

# Example: list all .txt files in current folder
files = glob.glob(os.path.join(".", "*.txt"))
print(files)

files=glob.glob("*nb")
print(files)


files=glob.glob(os.path.join(".","*nb"))
print(files)
```

```python
### C) argparse parses command line arguments (tiny demo)

import argparse

parser = argparse.ArgumentParser()
parser.add_argument("--limit", type=int, default=5)
args = parser.parse_args(args=["--limit", "12"])  # pretend command line input
print(args.limit)  # 12

## or 
# args = parser.parse_args() ## get it from command line
```
