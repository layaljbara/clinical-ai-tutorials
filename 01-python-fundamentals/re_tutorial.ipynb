
# Python `re` Tutorial: Regular Expressions

---

## 1. What is `re`?

`re` is Python’s built-in library for **regular expressions (regex)**.

Regex is Python’s way of saying:

> “Match text by pattern, not by exact characters.”

You use it to:

* search text
* extract information
* validate formats
* replace or clean strings

Examples: emails, dates, lab values, IDs, clinical notes, logs.

---

## 2. Importing the library

Example:

    import re

---

## 3. What is a regex pattern?

A **pattern** is a string that describes the *shape* of text.

Example:

    r"\\d+"

Meaning:

* `\\d` → any digit (0–9)
* `+` → one or more

So `\\d+` means: “one or more digits.”

---

## 4. What does `re.search()` mean?

### Core rule (very important)

> **`re.search()` finds the first match anywhere in the string.**

Example:

    import re
    text = "Patient age is 54 years"
    m = re.search(r"\\d+", text)
    print(m.group())

Output:

    54

What happens step by step:

1. Python scans the text from left to right
2. It finds the first place where the pattern matches
3. It returns a match object
4. `.group()` gives the matched text

If nothing is found, it returns `None`.

---

## 5. What does `re.findall()` mean?

### Core rule

> **`re.findall()` returns ALL matches as a list.**

Example:

    text = "AST 45 ALT 63 ALP 120"
    re.findall(r"\\d+", text)

Output:

    ['45', '63', '120']

Use this when you want **everything**, not just the first match.

---

## 6. What does `re.finditer()` mean?

> **`re.finditer()` returns matches with their positions.**

Example:

    text = "AST 45 ALT 63 ALP 120"
    for m in re.finditer(r"\\d+", text):
        print(m.group(), m.start(), m.end())

This is useful when you care about **where** something appears.

---

## 7. What does `re.match()` mean?

### Core rule

> **`re.match()` only checks the START of the string.**

Example:

    re.match(r"\\d+", "54 years")   # match
    re.match(r"\\d+", "age 54")     # None

This is often used for **validation**.

---

## 8. What does `re.sub()` mean?

### Core rule

> **`re.sub()` replaces all matches.**

Example:

    text = "Pt age: 54, id: 99102"
    re.sub(r"\\d+", "[NUM]", text)

Output:

    Pt age: [NUM], id: [NUM]

Used heavily for cleaning and anonymization.

---

## 9. What are groups?

Parentheses create **groups** — pieces you want to extract.

Example:

    text = "ALT 63 U/L"
    m = re.search(r"([A-Z]{2,5})\\s*(\\d+)", text)
    m.group(1)  # ALT
    m.group(2)  # 63

Groups turn messy text into **structured fields**.

---

## 10. What are flags?

Flags modify matching behavior.

Example:

    re.search(r"alt", "ALT is high", re.IGNORECASE)

Common flags:

* `re.IGNORECASE`
* `re.MULTILINE`
* `re.DOTALL`

---

## 11. Mini practical example

    text = "BP 120/80, HR 72 bpm, Temp 37.2C"
    pattern = r"([A-Za-z]+)\\s*([\\d./]+)"
    re.findall(pattern, text)

Output:

    [('BP', '120/80'), ('HR', '72'), ('Temp', '37.2')]

This shows how regex converts free text into structured data.

---

## 12. Golden rules (memorize these)

* `re.search()` → first match anywhere
* `re.match()` → must match the start
* `re.findall()` → all matches
* `re.finditer()` → matches + positions
* `re.sub()` → replace
* `(...)` → capture
* always use raw strings: `r"..."`

---

## 13. Final mental model

> 🔵 Pattern = describe what text LOOKS like  
> 🟢 Search/find = locate it  
> 🟠 Groups = extract it  
> 🔴 Sub = rewrite it

### Character Types

| Pattern | Meaning                        | Example | Matches             |
| ------- | ------------------------------ | ------- | ------------------- |
| `\d`    | digit                          | `\d`    | 0–9                 |
| `\D`    | not a digit                    | `\D`    | a, %, space         |
| `\w`    | word character                 | `\w`    | a–z, A–Z, 0–9, _    |
| `\W`    | not a word character           | `\W`    | @, space, -         |
| `\s`    | whitespace                     | `\s`    | space, tab, newline |
| `\S`    | not whitespace                 | `\S`    | letters, digits     |
| `.`     | any character (except newline) | `.`     | a, 3, @             |


### Quantifiers 
| Pattern | Meaning         | Example   | Matches       |
| ------- | --------------- | --------- | ------------- |
| `+`     | one or more     | `\d+`     | 1, 23, 456    |
| `*`     | zero or more    | `\d*`     | "", 1, 22     |
| `?`     | zero or one     | `colou?r` | color, colour |
| `{n}`   | exactly n       | `\d{3}`   | 123           |
| `{n,}`  | at least n      | `\d{2,}`  | 12, 1234      |
| `{n,m}` | between n and m | `\d{2,4}` | 12, 123, 1234 |


### Anchors (position)
| Pattern | Meaning           | Example   | Matches           |
| ------- | ----------------- | --------- | ----------------- |
| `^`     | start of string   | `^Hi`     | "Hi there"        |
| `$`     | end of string     | `end$`    | "the end"         |
| `\b`    | word boundary     | `\bcat\b` | "cat" not "catch" |
| `\B`    | not word boundary | `\Bcat`   | "scat"            |


### Groups and Logic

| Pattern | Meaning             | Example | Purpose |            |            |
| ------- | ------------------- | ------- | ------- | ---------- | ---------- |
| `( )`   | capture group       | `(\d+)` | extract |            |            |
| `(?: )` | non-capturing group | `(?:Dr  | Mr)`    | group only |            |
| `       | `                   | OR      | `cat    | dog`       | cat or dog |






### Lookarounds 

| Pattern    | Meaning             | Example      | Use                 |
| ---------- | ------------------- | ------------ | ------------------- |
| `(?=...)`  | positive lookahead  | `\d(?=mg)`   | digit before mg     |
| `(?!...)`  | negative lookahead  | `\d(?!mg)`   | digit not before mg |
| `(?<=...)` | positive lookbehind | `(?<=\$)\d+` | after $             |
| `(?<!...)` | negative lookbehind | `(?<!\$)\d+` | not after $         |


## Very common real patters

| Task               | Pattern                                          |
| ------------------ | ------------------------------------------------ |
| Number             | `\d+`                                            |
| Decimal            | `\d+\.?\d*`                                      |
| Date (YYYY-MM-DD)  | `\d{4}-\d{2}-\d{2}`                              |
| Email              | `[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}` |
| Word               | `[A-Za-z]+`                                      |
| Lab token          | `[A-Z]{2,5}\s*\d+`                               |
| Whitespace cleanup | `\s+`                                            |


Golden memory rules

• [] → choices

• () → extract

• + * ? {} → quantity

• ^ $ \b → position

• | → OR

• always use raw strings → r"...“
